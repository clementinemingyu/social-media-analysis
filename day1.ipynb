```python
import pandas as pd
df = pd.read_csv(
    "Capstone_Project_synthetic_social_media_data.csv"
)

df.head()
```

```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
df["UserGender"] = df["UserGender"].fillna(
    df["UserGender"].mode()[0]
)

df["SentimentScore"] = df["SentimentScore"].fillna(
    df["SentimentScore"].median()
)
df.isnull().sum()
```

```python
df["PostDateTime"] = pd.to_datetime(
    df["PostDateTime"],
    format="mixed"
)
df["PostDateTime"].dtype
```

```python
df.to_csv(
    "cleaned_social_media_data.csv",
    index=False
)
```
